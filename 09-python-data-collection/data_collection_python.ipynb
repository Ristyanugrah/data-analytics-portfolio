```python
from faker import Faker
import numpy as np

wadah_data = []
fake = Faker('id_ID')

for i in range(1000):
  student_id = np.random.randint(10000,20000)
  student_name = fake.name()

  subject_options = ['IPA','IPS','English','Math']
  subject = np.random.choice(subject_options)

  score = np.random.randint(0,101)

  satu_data_siswa = {
      'student_id': student_id,
      'student_name': student_name,
      'subject': subject,
      'score': score
  }

  wadah_data.append(satu_data_siswa)

print(wadah_data)
```

```python
import pandas as pd

student_table = pd.DataFrame(wadah_data)
student_table
```
