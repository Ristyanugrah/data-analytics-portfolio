```python
# Encapsulation
class Tabungan:

    # Constructor
    def __init__(self, id, nama, saldo):
        self.id = id
        self.nama = nama
        self._saldo = saldo

    # Method menambah saldo
    def deposit(self, jumlah):
        self._saldo += jumlah
        print("Saldo anda berhasil ditambahkan")
        print("Saldo anda sekarang :", self._saldo)

    # Method untuk mengambil saldo
    def tarik(self, jumlah):

        if jumlah > self._saldo:
            print("Saldo anda tidak cukup")
            print("Saldo anda sekarang :", self._saldo)

        else:
            self._saldo -= jumlah
            print("Saldo anda berhasil ditarik")
            print("Saldo anda sekarang :", self._saldo)


# Object
tabungan1 = Tabungan(1, "Risty", 5000000)

# Input jumlah tarik dari user
jumlah_tarik = int(input("Masukkan jumlah yang ingin ditarik: "))

# Panggil method tarik
tabungan1.tarik(jumlah_tarik)
```

```python
class Pegawai:

    def __init__(self, nama, gaji):
        self.nama = nama
        self.gaji = gaji


class Manager(Pegawai):

    def __init__(self, nama, gaji, tunjangan):
        super().__init__(nama, gaji)
        self.tunjangan = tunjangan

    def info_manager(self):
        print("Nama :", self.nama)
        print("Gaji :", self.gaji)
        print("Tunjangan :", self.tunjangan)


manager1 = Manager("Risty Anugrah", 7000000, 2000000)

manager1.info_manager()
```

```python
class Kendaraan:

    def bahan_bakar(self):
        print("Jenis bahan bakar tidak diketahui")


class Mobil(Kendaraan):

    def bahan_bakar(self):
        print("Bensin")


class Motor(Kendaraan):

    def bahan_bakar(self):
        print("Solar")


mobil1 = Mobil()
motor1 = Motor()

mobil1.bahan_bakar()
motor1.bahan_bakar()
```

```python
from abc import ABC, abstractmethod


class Hewan(ABC):

    @abstractmethod
    def suara(self):
        pass


class Kuda(Hewan):

    def suara(self):
        print("whee")


class kerbau(Hewan):

    def suara(self):
        print("Mooo")


kuda1 = Kuda()
kerbau1 = kerbau()

kuda1.suara()
kerbau1.suara()
```

**Jawaban Nomer 1**
Konsep OOP penting karena membantu pengembang mengatur kode menjadi lebih terstruktur dan mudah dikelola. Class digunakan untuk membuat objek yang memiliki data dan fungsi tertentu. Encapsulation membantu melindungi data agar tidak dapat diubah secara sembarangan. Inheritance memungkinkan penggunaan kembali kode yang sudah ada sehingga mengurangi penulisan kode yang berulang. Polymorphism memungkinkan penggunaan metode yang sama dengan perilaku yang berbeda pada setiap objek. Abstraction membantu menyembunyikan detail implementasi yang tidak perlu diketahui pengguna. Dengan menerapkan konsep-konsep tersebut, aplikasi menjadi lebih rapi, mudah dikembangkan, dan lebih mudah dipelihara dalam jangka panjang.
