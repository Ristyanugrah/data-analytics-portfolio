## Soal 1.A Kalkulator Sederhana

Buat program kalkulator yang meminta dua angka dan operator (+, -, *, /), lalu tampilkan hasilnya

```python
angka_1 = float(input("Masukkan angka pertama: "))
angka_2 = float(input("Masukkan angka kedua: "))
operator = input("Masukkan operator: ")

if operator == "+":
  hasil = angka_1 + angka_2
elif operator == "-":
  hasil = angka_1 - angka_2
elif operator == "*":
  hasil = angka_1 * angka_2
elif operator == "/":
  if angka_2 == 0:
    print("Sesuatu yg dibagi 0. Pembagi itu gaboleh 0")
  else:
    hasil = angka_1 / angka_2
else:
  print("Operator tidak diketahui!")

if operator in ("+","-","*") or (operator == "/" and angka_2 != 0):
  print(f"Hasil:  {angka_1} {operator} {angka_2} = {hasil}")
```

## Soal 2.A Analisis Data Penjualan dengan List


Diberikan data penjualan 7 hari dalam sebuah list. Hitung total, rata-rata, temukan nilai penjualan terbaik dan terburuk, serta hari mana yang paling tinggi.

```python
hari = ["Senin","Selasa","Rabu","Kamis","Jumat","Sabtu","Minggu"]
penjualan = [100000, 5000, 60000, 10000000, 900000, 1000, 350000]

# kalkulasi dasarnya inituhh
total = sum(penjualan)
rata_rata = total / len(penjualan)
max_penjualan = max(penjualan)
min_penjualan = min(penjualan)

# temuin hari dg menemukan index penjualan terbaik dan terburuknya
index_penjualan_terbaik = penjualan.index(max_penjualan)
index_penjualan_terendah = penjualan.index(min_penjualan)

# print(penjualan_terbaik)
# print(penjualan_terendah)

# hari terbaik dan hari terburuk
hari_terbaik = hari[index_penjualan_terbaik]
hari_terendah = hari[index_penjualan_terendah]

print(f"{'='*40}")
print("LAPORAN PENJUALAN MINGGUAN")
print(f"{'='*40}")
print(f"Total Penjualan: Rp {total: ,.0f} ")
print(f"Rata-rata / hari: Rp {rata_rata: ,.2f}")
print(f"Hari dengan penjualan tertinggi: {hari_terbaik}")
print(f"Hari dengan penjualan terendah: {hari_terendah}")
print(f"Penjualan tertinggi: Rp{max_penjualan:,.0f} ")
print(f"Penjualan terendah:  Rp{min_penjualan:,.0f} ")

print("Hari dengan penjualan tertinggi: " , hari_terbaik)
```

## Soal 2.B Manajemen Data Siswa dengan Dictionary

Buat sistem sederhana menyimpan data satu siswa menggunakan dictionary, lalu tambahkan nilai rata-rata dan grade-nya ke dictionary tersebut.

buat 1 dictionary
-> yg isinya nama, sama nilai

dari nilai2 ini, kita itung rata2 nilainya berapa, dan  gradenya apa

-> rata-rata
-> grade
ini kita masukin kembali ke dalam dictionary

```python
siswa = {
    "nama": "Risty Anugrah",
    "nilai_mtk": 100,
    "nilai_ipa": 96,
    "nilai_olahraga": 55
}

# hitung nilai rata-rata
total_nilai = siswa["nilai_mtk"] + siswa["nilai_ipa"] + siswa["nilai_olahraga"]
rata_rata_nilai = total_nilai / 3

# masukkin rata-rata ke dalam dictionary
siswa["rata_rata"] = round(rata_rata_nilai,2)

# tentuin grade
if rata_rata_nilai >= 90:
  siswa["grade"] = "A"
elif rata_rata_nilai >= 75:
  siswa["grade"] = "B"
elif rata_rata_nilai >= 60:
  siswa["grade"] = "C"
else:
  siswa["grade"] = "D"


# print(siswa)

# tampilkan
print(f"{'='*40}")
print("Laporan Nilai Siswa DABI23")
print(f"{'='*40}")
print(f"Nama:  {siswa["nama"]} ")
print(f"Rata-rata: {siswa["rata_rata"]} ")
print(f"Grade:  {siswa["grade"]} ")
```

```python
#belanjar cuyy di indomaret

keranjang = {
    "item_1": {"nama": "Popmie", "harga": 65000, "qty": 5},
    "item_2": {"nama": "Nescafe", "harga": 150000, "qty": 1},
    "item_3": {"nama": "Chitato", "harga": 135000, "qty": 2}
}

nama_pembeli = input("Masukkan nama pembeli: ")

# itung total belanja
total_item_1 = keranjang["item_1"]["harga"] * keranjang["item_1"]["qty"]
total_item_2 = keranjang["item_2"]["harga"] * keranjang["item_2"]["qty"]
total_item_3 = keranjang["item_3"]["harga"] * keranjang["item_3"]["qty"]

total_sebelum_diskon = total_item_1 + total_item_2 + total_item_3

if total_sebelum_diskon >= 1000000 and keranjang["item_1"]["nama"] == "Popmie" :
  diskon = 2
  note = "Diskon 10%"
elif total_sebelum_diskon >= 5000000:
  diskon = 1
  note = "Diskon kamu cuma 5% tapi okelah"
else:
  diskon = 0
  note = "Kamu ga dapet, makanya belanja lagi yaa"

harga_diskon = total_sebelum_diskon * diskon
total_setelah_diskon = total_sebelum_diskon - harga_diskon

# tampilin struk
print(f"{'='*40}")
print("DABI MART")
print(f"{'='*40}")

print(f"Pembeli: {nama_pembeli}")
print("DABI MART")
print(f"{'='*40}")

print("List Belanja")
print(f"{keranjang['item_1']['nama']} = {total_item_1} ")
print(f"{keranjang['item_2']['nama']} = {total_item_2} ")
print(f"{keranjang['item_3']['nama']} = {total_item_3} ")
print(f"Harga Sebelum Diskon:  {total_sebelum_diskon} ")
print(f"Total Diskon: {harga_diskon} ")

print(f"{'='*40}")
print(f"TOTAL BAYAR: {total_setelah_diskon}")
```

## Soal 1 Konverter Suhu

Buat program Konverter Suhu. Program meminta input nilai suhu
dalam Celsius, lalu tampilkan konversinya ke Fahrenheit (F = C × 9/5 + 32 ) dan Kelvin (K = C + 273.15 )

```python
# Input suhu dalam Celsius
celsius = float(input("Masukkan suhu dalam Celsius: "))

# Konversi
fahrenheit = celsius * 9/5 + 32
kelvin = celsius + 273.15

# Output hasil
print("\nHasil Konversi:")
print(f"Celsius    : {celsius}°C")
print(f"Fahrenheit : {fahrenheit:.2f}°F")
print(f"Kelvin     : {kelvin:.2f} K")
```

## Soal 2 Manajemen Data Kota

Buat dictionary yang menyimpan data 3 kota (nama kota, provinsi,
jumlah penduduk). Dari 3 kota tersebut, tampilkan: total penduduk ketiganya, rata-rata penduduk, dan kota mana yang penduduknya paling banyak (gunakan if/elif/else).

```python
# Data 3 kota
kota = {
    "kota_1": {"nama": "Jakarta", "provinsi": "DKI Jakarta", "penduduk": 10600000},
    "kota_2": {"nama": "Bandung", "provinsi": "Jawa Barat", "penduduk": 2500000},
    "kota_3": {"nama": "Surabaya", "provinsi": "Jawa Timur", "penduduk": 2900000},
}

# Ambil jumlah penduduk
p1 = kota["kota_1"]["penduduk"]
p2 = kota["kota_2"]["penduduk"]
p3 = kota["kota_3"]["penduduk"]

# Total & rata-rata
total_penduduk = p1 + p2 + p3
rata_rata = total_penduduk / 3

# Cari kota dengan penduduk terbanyak (pakai if/elif/else)
if p1 > p2 and p1 > p3:
    kota_terbesar = kota["kota_1"]["nama"]
elif p2 > p1 and p2 > p3:
    kota_terbesar = kota["kota_2"]["nama"]
else:
    kota_terbesar = kota["kota_3"]["nama"]

# Output
print("=== DATA KOTA ===")
print(f"Total penduduk : {total_penduduk:,}")
print(f"Rata-rata      : {rata_rata:,.0f}")
print(f"Kota terbesar  : {kota_terbesar}")
```

## Soal 3

Buat program pengecekan kelayakan pinjaman. Program meminta:
umur, penghasilan bulanan, dan jumlah pinjaman yang diajukan. Tentukan apakah pengajuan disetujui/ditolak berdasarkan aturan: umur 21-55 tahun, penghasilan minimal Rp 5.000.000, dan cicilan perkiraan (pinjaman dibagi 12) tidak lebih dari 30% penghasilan.

```python
# Input data
umur = int(input("Masukkan umur: "))
penghasilan = float(input("Masukkan penghasilan bulanan (Rp): "))
pinjaman = float(input("Masukkan jumlah pinjaman (Rp): "))

# Hitung cicilan per bulan (asumsi 12 bulan)
cicilan = pinjaman / 12

# Cek syarat
if 21 <= umur <= 55:
    if penghasilan >= 5000000:
        if cicilan <= 0.3 * penghasilan:
            status = "Disetujui"
        else:
            status = "Ditolak (cicilan terlalu besar)"
    else:
        status = "Ditolak (penghasilan kurang)"
else:
    status = "Ditolak (umur tidak memenuhi syarat)"

# Output
print("\n=== HASIL PENILAIAN ===")
print(f"Umur            : {umur} tahun")
print(f"Penghasilan     : Rp {penghasilan:,.0f}")
print(f"Pinjaman        : Rp {pinjaman:,.0f}")
print(f"Cicilan/bulan   : Rp {cicilan:,.0f}")
print(f"Status          : {status}")
```
