## 📚 Tuple

Tuple adalah tipe data yang digunakan untuk menyimpan banyak nilai dalam satu variabel.  
Tuple termasuk struktur data linear, artinya:

- Data tersusun berurutan  
- Setiap elemen memiliki posisi (index)  

Tuple ditulis menggunakan tanda kurung biasa `( )`

### 🔑 Karakteristik Tuple

Data dalam tuple memiliki urutan yang tetap.


```python

campuran = (1, "hello", 3.5, True)
```


#### 6. Bisa di-iterasi (looping)
```python
data = (10, 20, 30)
for item in data:

```

#### 7. Fixed Size
Tuple tidak bisa bertambah atau berkurang.

## 📚 Set

Set adalah tipe data yang digunakan untuk menyimpan banyak nilai dalam satu variabel. Set termasuk struktur data non-linear, artinya:

- Data tidak memiliki urutan (unordered)  
- Tidak memiliki index (tidak bisa diakses dengan posisi)  

Set ditulis menggunakan tanda kurung kurawal `{ }`

---

### 🔑 Karakteristik Set

#### 1. Unordered (tidak berurutan)
Data dalam set tidak memiliki urutan tetap.
```python
data = {1, 2, 3}
```

#### 2. Unindexed (tidak memiliki index)
Set tidak bisa diakses menggunakan index.
```python
data = {10, 20, 30}
print(data[0]) ❌ error
```

#### 3. No Duplicates (tidak boleh duplikat)
Set hanya menyimpan nilai unik.
```python
data = {1, 1, 2, 3}
print(data)  # {1, 2, 3}
```

#### 4. Mutable (bisa diubah)
Set bisa ditambah atau dihapus elemennya.
```python
data = {1, 2, 3}
data.add(4)
data.remove(2)
```

#### 5. Bisa menyimpan berbagai tipe data (terbatas)
Set bisa menyimpan berbagai tipe data, tapi harus immutable.
```python
data = {1, "hello", 3.5}
```

#### 6. Bisa di-iterasi (looping)
```python
data = {10, 20, 30}
for item in data:
    print(item)
```

#### 7. Tidak bisa menyimpan mutable object
List atau dictionary tidak bisa dimasukkan ke dalam set.
```python
data = {1, 2, 3}
data.add([4,5]) ❌ error
```

## 📚 Dictionary

Dictionary adalah tipe data yang digunakan untuk menyimpan data dalam bentuk pasangan key–value. Dictionary memungkinkan kita mengakses nilai berdasarkan key, bukan berdasarkan index.

Dictionary ditulis menggunakan tanda kurung kurawal `{ }` dengan format `key: value`.

---

### 🔑 Karakteristik Dictionary

#### 1. Key–Value Structure
Setiap data terdiri dari pasangan key dan value.
```python
data = {"nama": "Alice", "umur": 25}
```

#### 2. Ordered (Python 3.7+)
Dictionary mempertahankan urutan data sesuai urutan input.
```python
data = {"a": 1, "b": 2, "c": 3}
```

#### 3. Mutable (bisa diubah)
Dictionary dapat diubah setelah dibuat.
```python
data = {"nama": "Alice"}
data["nama"] = "Bob"
print(data)  # {'nama': 'Bob'}
```

#### 4. Key harus unik
Key tidak boleh duplikat, tetapi value boleh sama.
```python
data = {"a": 1, "a": 2}
print(data)  # {'a': 2}
```

#### 5. Bisa menyimpan berbagai tipe data
Value dalam dictionary bisa berbagai tipe data.
```python
data = {"nama": "Alice", "umur": 25, "aktif": True}
```

#### 6. Diakses menggunakan key
Data diakses menggunakan key, bukan index.
```python
data = {"nama": "Alice", "umur": 25}
print(data["nama"])  # Alice
```

#### 7. Dynamic Size
Dictionary bisa bertambah atau berkurang.
```python
data = {"a": 1}
data["b"] = 2
del data["a"]
```

**Soal 4 (Tantangan)**: Gunakan NumPy untuk membuat array berisi 6 data penjualan. Hitung mean, median, nilai maks & min menggunakan fungsi NumPy. Cek apakah nilai maks lebih dari dua kali lipat nilai min.

`````python
furniture = ["sofa","meja makan","tv"]
harga_furniture = [5000000,1000000,3500000]

total_harga = sum(harga_furniture)
rata_rata = round(total_harga / len(harga_furniture))

harga_tertinggi = max(harga_furniture)
harga_terendah = min(harga_furniture)

print("DAFTAR PRODUCT", furniture)
print("DAFTAR HARGA", harga_furniture)
print("\n")
print(f"Total Harga:  {total_harga}")
print(f"Raga-rata Harga: {rata_rata} ")
print(f"Harga Tertinggi: {harga_tertinggi}")
print(f"Harga Terendah: {harga_terendah}")
`````

Soal 3: Buat dictionary untuk menyimpan data 1 siswa (nama, nilai UTS, nilai UAS, nilai tugas). Hitung nilai akhir dengan bobot: UTS 30%, UAS 40%, tugas 30%. Tampilkan apakah siswa tersebut lulus (nilai akhir ≥ 70) atau tidak.

`````python
siswa = {
    "nama" : "Risty Anugrah",
    "uts" : 90,
    "uas" : 80,
    "tugas":100
}

nilai_akhir = (
  (siswa["uts"] *0.3)+
  (siswa["uas"] *0.4)+
  (siswa["tugas"] *0.3)
)

if nilai_akhir >= 70:
   print("lulus")
else:
   print("Tidak lulus")
`````
