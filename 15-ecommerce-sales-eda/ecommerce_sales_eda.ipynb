```python
# ==========================================================
# 5.1 FEATURE ENGINEERING
# Mengubah InvoiceDate menjadi format datetime
# ==========================================================

df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

print(df['InvoiceDate'].dtype)
```

```python
# ==========================================================
# Membuat Kolom Year
# ==========================================================

df['Year'] = df['InvoiceDate'].dt.year

df[['InvoiceDate','Year']].head()
```

```python
# ==========================================================
# Membuat Kolom Month
# ==========================================================

df['Month'] = df['InvoiceDate'].dt.month_name()

df[['InvoiceDate','Month']].head()
```

```python
# ==========================================================
# Membuat Nomor Bulan
# ==========================================================

df['MonthNumber'] = df['InvoiceDate'].dt.month

df[['InvoiceDate','MonthNumber']].head()
```

```python
# ==========================================================
# Membuat Nama Hari
# ==========================================================

df['DayName'] = df['InvoiceDate'].dt.day_name()

df[['InvoiceDate','DayName']].head()
```

```python
# ==========================================================
# Membuat Quarter
# ==========================================================

df['Quarter'] = df['InvoiceDate'].dt.quarter

df[['InvoiceDate','Quarter']].head()
```

```python
# ==========================================================
# Membuat Revenue
# ==========================================================

df['Revenue'] = df['Quantity'] * df['UnitPrice']

df[['Quantity','UnitPrice','Revenue']].head()
```

Business Question 1


Produk apa yang memberikan total pendapatan (Revenue) terbesar?
Tujuan Bisnis

Mengetahui produk yang menjadi penyumbang pendapatan terbesar sehingga perusahaan dapat memprioritaskan stok, promosi, dan strategi pemasaran.[teks link](https://)

```python
# ==========================================================
# BUSINESS QUESTION 1
# Produk dengan Revenue Terbesar
# ==========================================================

top_product = (
    df.groupby("Description")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_product
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=top_product.values,
    y=top_product.index,
    palette="viridis"
)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Total Revenue")
plt.ylabel("Product")

plt.show()
```

Berdasarkan hasil analisis, beberapa produk memberikan kontribusi revenue paling besar dibandingkan produk lainnya. Produk-produk tersebut merupakan produk unggulan perusahaan sehingga perlu dipastikan ketersediaan stoknya serta diprioritaskan dalam strategi promosi untuk meningkatkan pendapatan.

Business Question 2

Negara mana yang menghasilkan revenue terbesar?
Tujuan Bisnis

Menentukan pasar utama perusahaan.

Metode

✅ Group By

✅ Sorting

```python
country = (
    df.groupby("Country")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

country
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=country.values,
    y=country.index,
    palette="crest"
)

plt.title("Top 10 Countries by Revenue")

plt.xlabel("Revenue")

plt.ylabel("Country")

plt.show()
```

Negara dengan revenue terbesar merupakan pasar utama perusahaan. Perusahaan dapat memfokuskan strategi pemasaran, distribusi, dan pengembangan bisnis pada negara-negara tersebut untuk mempertahankan maupun meningkatkan penjualan.

Business Question 3

Bagaimana tren revenue setiap bulan?
Tujuan Bisnis

Melihat pola penjualan sepanjang waktu.

```python
monthly = (
    df.groupby("Month")["Revenue"]
      .sum()
      .reset_index()
)

monthly
```

```python
plt.figure(figsize=(12,6))

sns.lineplot(
    data=monthly,
    x="Month",
    y="Revenue",
    marker="o",
    linewidth=3
)

plt.title("Monthly Revenue Trend")

plt.xlabel("Month")

plt.ylabel("Revenue")

plt.show()
```

Revenue mengalami fluktuasi pada setiap bulan. Bulan dengan revenue tertinggi dapat dijadikan acuan dalam menentukan periode promosi, sedangkan bulan dengan revenue rendah dapat dievaluasi melalui strategi pemasaran atau pemberian diskon.

Business Question 4


Siapa pelanggan yang memiliki nilai transaksi terbesar?
Tujuan Bisnis

Mengidentifikasi pelanggan loyal.

```python
customer = (
    df.groupby("CustomerID")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

customer
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=customer.values,
    y=customer.index.astype(str),
    palette="rocket"
)

plt.title("Top 10 Customers by Revenue")

plt.xlabel("Revenue")

plt.ylabel("Customer ID")

plt.show()
```

Pelanggan dengan total pembelian terbesar merupakan pelanggan bernilai tinggi (high-value customers). Perusahaan dapat mempertahankan pelanggan tersebut melalui program loyalitas, diskon khusus, maupun penawaran eksklusif.

Business Question 5


Bagaimana total revenue pada setiap negara di setiap bulan?

Ini pertanyaan yang lebih profesional karena menggunakan Pivot Table.

Tujuan Bisnis

Membandingkan performa penjualan antar negara berdasarkan bulan.

```python
pivot = pd.pivot_table(
    df,
    index="Country",
    columns="Month",
    values="Revenue",
    aggfunc="sum",
    fill_value=0
)

pivot.head()
```

```python
plt.figure(figsize=(15,8))

sns.heatmap(
    pivot,
    cmap="YlGnBu"
)

plt.title("Revenue by Country and Month")

plt.show()
```

Heatmap menunjukkan bahwa setiap negara memiliki pola penjualan yang berbeda pada setiap bulan. Negara dengan warna yang lebih gelap menunjukkan kontribusi revenue yang lebih tinggi. Informasi ini dapat dimanfaatkan untuk menentukan waktu promosi yang paling efektif pada masing-masing pasar.
