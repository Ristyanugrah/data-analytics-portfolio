Diberikan sebuah list data mentah yang berisi campuran angka & teks (data sampah) seperti berikut :
raw_data = [120, -5, "ERROR", 250, 0, "NULL", 420, -10, 500]
Dari data di atas, buatlah sebuah fungsi bernama clean_data yang  menggunakan ‘for ‘loop untuk mengambil hanya angka yang memiliki nilai >= 200
hint : untuk mengecek data type setiap value, gunakan isinstance(value, type_data)


```python
raw_data = [120, -5, "ERROR", 250, 0, "NULL", 420, -10, 500]

def clean_data(data):

    cleaned = []

    for value in data:

        if isinstance(value, int):

            if value >= 200:

                cleaned.append(value)

    return cleaned


result = clean_data(raw_data)

print(result)
```

Dari data tersebut menggunakan list comprehension, buatlah sebuah list baru bernama price_with_tax yang berisikan nilai yang telah dikalikan oleh 11%. Lalu dari hasil data price_with_tax , buatlah variable baru roundup_price_tax dan gunakan lambda function untuk mengubah nilainya agar memiliki nilai hanya 2 angka di belakang koma.


```python
price_with_tax = [price * 1.11 for price in result]

print(price_with_tax)

roundup_price_tax = list(
    map(lambda x: round(x, 2), price_with_tax)
)
print(roundup_price_tax)
```

Gunakan nested loop untuk membuat tabel perkalian 10 × 10 agar menghasilkan output

```python
for i in range(1, 11):

    for j in range(1, 11):

        print(f"{i*j:4}", end=" ")

    print()
```
